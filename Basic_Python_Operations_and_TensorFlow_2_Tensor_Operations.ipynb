What is Python?

Python is a high-level, interpreted, general-purpose programming language. It is widely used in Artificial Intelligence, Machine Learning, Data Science, and Generative AI because of its simple syntax and extensive libraries.

**What is TensorFlow?**

TensorFlow is an open-source machine learning framework developed by Google. It is used to build, train, and deploy machine learning and deep learning models.

TensorFlow represents data using tensors.

A tensor is a multidimensional array containing numerical values.

For example:

Scalar → 5

Vector → [1, 2, 3]

Matrix →
        [1 2]
        [3 4]
        
3-D Tensor → multiple matrices

What is a Tensor?

A tensor is a generalization of scalars, vectors, and matrices.

```python
import tensorflow as tf

print("TensorFlow Version:", tf.__version__)
```

```python
a = tf.constant([[1, 2, 3],
                 [4, 5, 6]], dtype=tf.float32)

b = tf.constant([[6, 5, 4],
                 [3, 2, 1]], dtype=tf.float32)

print("Tensor A:")
print(a.numpy())

print("\nTensor B:")
print(b.numpy())
```

```python
print("Addition:")
print(tf.add(a, b).numpy())
```

```python
print("Subtraction:")
print(tf.subtract(a, b).numpy())
```

```python
print("Multiplication:")
print(tf.multiply(a, b).numpy())
```

```python
print("Division:")
print(tf.divide(a, b).numpy())
```

```python
print("Matrix Multiplication:")
print(tf.matmul(a, tf.transpose(b)).numpy())
```

```python
print("Transpose:")
print(tf.transpose(a).numpy())
```

```python
x = tf.constant([1, 2, 3, 4, 5, 6])

print("Original:")
print(x.numpy())

print("Reshaped:")
print(tf.reshape(x, [2, 3]).numpy())
```

```python
print("Stack:")
print(tf.stack([a, b]).numpy())
```

```python
import tensorflow as tf

print("TensorFlow Version:", tf.__version__)

# Create tensors
a = tf.constant([[1, 2, 3],
                 [4, 5, 6]], dtype=tf.float32)

b = tf.constant([[6, 5, 4],
                 [3, 2, 1]], dtype=tf.float32)

print("\nTensor A:")
print(a.numpy())

print("\nTensor B:")
print(b.numpy())

# Arithmetic operations
print("\nAddition:")
print(tf.add(a, b).numpy())

print("\nSubtraction:")
print(tf.subtract(a, b).numpy())

print("\nMultiplication:")
print(tf.multiply(a, b).numpy())

print("\nDivision:")
print(tf.divide(a, b).numpy())

# Matrix operations
print("\nMatrix Multiplication:")
print(tf.matmul(a, tf.transpose(b)).numpy())

print("\nTranspose:")
print(tf.transpose(a).numpy())

# Reshape
x = tf.constant([1, 2, 3, 4, 5, 6])
print("\nReshaped Tensor:")
print(tf.reshape(x, [2, 3]).numpy())

# Reduction operations
print("\nSum:", tf.reduce_sum(a).numpy())
print("Mean:", tf.reduce_mean(a).numpy())
print("Maximum:", tf.reduce_max(a).numpy())
print("Minimum:", tf.reduce_min(a).numpy())

# Concatenation
print("\nConcatenation:")
print(tf.concat([a, b], axis=0).numpy())

# Stack
print("\nStack:")
print(tf.stack([a, b]).numpy())

# Activation functions
x = tf.constant([-2.0, -1.0, 0.0, 1.0, 2.0])

print("\nReLU:")
print(tf.nn.relu(x).numpy())

print("\nSigmoid:")
print(tf.nn.sigmoid(x).numpy())

print("\nTanh:")
print(tf.nn.tanh(x).numpy())

print("\nSoftmax:")
print(tf.nn.softmax(x).numpy())

# Tensor properties
print("\nShape:", a.shape)
print("Data Type:", a.dtype)
print("Rank:", tf.rank(a).numpy())
```
